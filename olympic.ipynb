[1] make sure that all the columns has the word "in" in them

```python
# number of records that do not have "in" in it without Null values --> 22820
df[~df["Born"].str.contains("in", na=True)]["Born"].count()
```

```python
Born_Pattern = r"(\d+ \w+ \d+)"
len(df[~df["Born"].str.contains("in", na=True) & ~df["Born"].str.match(Born_Pattern, na=True)])
len(df[~df["Born"].str.contains("in", na=True) & ~df["Born"].str.match(Born_Pattern, na=True)])
```

[2] replace the "in" and "(" with "," and the ")" with ""

```python
df["Born"] = df["Born"].str.replace("in", ",", n=1)
df["Born"] = df["Born"].str.replace("(", ",")
df["Born"] = df["Born"].str.replace(")", "")

df.head()
```

[3] split the column into Born date, Born city, Born region, Born country

```python
df[["Born date", "Born city", "Born region", "Born country"]] = df["Born"].str.split(",", n=3, expand=True)
df.head()
```

```python
import numpy as np
df["Born date"] = df["Born date"].replace("", np.nan)
```

```python
new_born_date_pattern = r"(\D+ \W+ \D+)"
df[~df["Born date"].str.match(new_born_date_pattern, na=True) & ~df["Born date"].str.match(r"(\d)", na=True)].head()
```

```python
len(df[df["Born city"].str.match(r"\w+ \d+", na=False)]) # 50
df[df["Born city"].str.match(r"\w+ \d+", na=False)].head()
```

```python
# fill the NON values in the Born date column which has it's date in the Born city (c. |  circa)
df["Born date"] = df["Born date"].fillna(df[df["Born city"].str.match(r"(\w+)", na=False)]["Born city"].str.extract(r"(\d+)").squeeze())
```

```python
df.loc[df["Born city"].str.contains("c.", na=False), "Born city"] = None
df.loc[df["Born city"].str.contains("circa", na=False), "Born city"] = None
df.loc[df["Born city"].str.contains("1926 or 1927", na=False), "Born city"] = None
```

```python
len(df[df["Born city"].str.match(r"\w+", na=False)]) # 53
df[df["Born city"].str.match(r"\w+", na=False)].head()
```

```python
df["Born date"] = pd.to_datetime(df["Born date"], errors="coerce")
df.head()
```
